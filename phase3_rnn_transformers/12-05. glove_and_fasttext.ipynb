```python
import urllib.request
import zipfile
from lxml import etree
import re
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
# Download data
urllib.request.urlretrieve("https://raw.githubusercontent.com/ukairia777/tensorflow-nlp-tutorial/main/09.%20Word%20Embedding/dataset/ted_en-20160408.xml", filename="ted_en-20160408.xml")
```

```python
targetXML = open('ted_en-20160408.xml', 'r', encoding='UTF8')
target_text = etree.parse(targetXML)

# Extract only the text between <content> and </content> from the XML file.
parse_text = '\n'.join(target_text.xpath('//content/text()'))

# Use regex substitution to remove background-sound tags such as (Audio) and (Laughter) inside the content.
# This code removes text enclosed in parentheses.
content_text = re.sub(r'\([^)]*\)', '', parse_text)

# Perform sentence tokenization on the input corpus using NLTK.
sent_text = sent_tokenize(content_text)

# For each sentence, remove punctuation and convert uppercase letters to lowercase.
normalized_text = []
for string in sent_text:
     tokens = re.sub(r"[^a-z0-9]+", " ", string.lower())
     normalized_text.append(tokens)

# Perform word tokenization for each sentence using NLTK.
result = [word_tokenize(sentence) for sentence in normalized_text]
```

```python
print('Total number of samples: {}'.format(len(result)))
```

```python
# Print only 3 samples
for line in result[:3]:
    print(line)
```

```python
from gensim.models import Word2Vec
from gensim.models import KeyedVectors
```

```python
model = Word2Vec(sentences=result, vector_size=100, window=5, min_count=5, workers=4, sg=0)
```

Here are the Word2Vec arguments used in this notebook.

* size = dimensionality of the word vectors (embedding dimension)
* window = context window size
* min_count = minimum word frequency threshold (words below this are ignored)
* workers = number of worker processes for training
* sg = 0 for CBOW, 1 for Skip-gram

```python
model_result = model.wv.most_similar("man")
print(model_result)
```

```python
model.wv.save_word2vec_format('eng_w2v') # Save the model
loaded_model = KeyedVectors.load_word2vec_format("eng_w2v") # Load the model
```

```python
model_result = loaded_model.most_similar("man")
print(model_result)
```
